```
import torch

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', device)
torch.manual_seed(42)
```

```
# Creation, indexing, reshaping, matrix multiplication, broadcasting and aggregation
tensor = torch.arange(12, dtype=torch.float32).reshape(3, 4)
print('Tensor:\n', tensor)
print('Indexed row:', tensor[1])
print('Reshaped:', tensor.reshape(2, 6))
weights = torch.randn(4, 2)
print('Matrix multiplication:', tensor @ weights)
print('Broadcasting:', tensor + torch.tensor([1., 2., 3., 4.]))
print('Column means:', tensor.mean(dim=0))
gpu_tensor = tensor.to(device)
print('Tensor device:', gpu_tensor.device)
```
